```python
cat_feature=['person_education','person_gender','person_emp_exp','person_home_ownership','loan_intent','previous_loan_defaults_on_file']
```

```python
from scipy.stats import chi2_contingency


alpha=0.05
#df_cleaned['charges_bin']=pd.qcut(df_cleaned['charges'],q=4,labels=False)
chi2_result={}

for col in cat_feature:
    contingency=pd.crosstab(df[col],df['loan_status'])
    chi2_stat,p_val,_,_=chi2_contingency(contingency)
    decision="Reject Null (Keep Feature)" if p_val <alpha else"Accept Null (Drop Feature)"
    chi2_result[col]={
        "chi2_stat":chi2_stat,
        'p_value':p_val,
        "Decision":decision
    }
```

```python
chi2_df=pd.DataFrame(chi2_result).T
chi2_df=chi2_df.sort_values(by='p_value')
chi2_df
```

```python
df.drop(['person_education','cb_person_cred_hist_length'],inplace=True,axis=1)
```
